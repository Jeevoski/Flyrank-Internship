## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
Task Type: Binary Classification combined with Ranking (Opportunity Scoring).
Explanation: The system acts as a proactive monitoring filter. We first classify whether a page is exhibiting signals of algorithmic decay (binary: yes/no), and then output a probability score to rank those pages so the operations team knows which one to review first
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
The Target: Predicting whether a high-value page will continue to lose traffic over the next 30 days without intervention.
The Proxy (for this starter data): Because the starter data is a single snapshot, we will use a proxy label: is_declining_label = (trend_direction == "down"). We are trying to predict this current state using historical engagement and position features.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
Metric: Precision@50 (or Precision@K).
Why: Overall accuracy does not matter because the operations team only has the capacity to review and rewrite a limited number of pages per week (e.g., 50 pages). We need to measure how many of the top 50 pages the model recommends are actually decaying and require intervention.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
The Unit of Analysis: One row equals one distinct content page (represented by content_id), evaluated over a 90-day window. The dataframe below demonstrates this structure along with our proxy target column (is_declining_label).
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
The limitation of rules: A fixed rule (like "flag if impressions > 500 and trend is down") is too rigid. It treats a 6-month-old page at position 3 the exact same as a 3-year-old page at position 8, resulting in a massive queue of false positives.
The ML advantage: Machine learning can weigh complex, non-linear interactions—such as how age, engagement rate, and click-through-rate degradation behave together—to find subtle, true decay patterns that rigid logic gates miss.
```
