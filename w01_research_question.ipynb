## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
The Question: Can we accurately identify which high-value pages are entering a state of algorithmic decay based on their historical engagement and position trends?

Unit of Analysis: A single web page.

The Decision: Should the content operations team allocate manual labor hours to update this specific page this week?

The Action: Dispatching an editor to rewrite outdated sections and improve the page's relevance to current search intents.

Cost of a Wrong Call: A False Positive creates logistical waste by burning budget and labor hours on a page that did not actually need intervention. A False Negative results in a decaying page dropping out of top rankings, leading to a compounded loss of organic traffic that is highly expensive to win back.

Why ML helps: A hand-written rule (e.g., "refresh if traffic drops 10%") is too rigid and flags too many false positives. Machine learning can weigh multiple complex variables simultaneously to isolate true, subtle signals of decay.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset directly from your public GitHub repository
csv_url = 'https://raw.githubusercontent.com/Jeevoski/Flyrank-Internship/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(csv_url)

# Calculate dataset scale and baseline decay rate
total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
pct_declining = (declining_pages / total_pages) * 100

# Calculate age discrepancies to prove there is observable signal
avg_age_declining = df[df['trend_direction'] == 'down']['content_age_days'].mean()
avg_age_stable = df[df['trend_direction'] != 'down']['content_age_days'].mean()

print(f"Total pages monitored: {total_pages:,}")
print(f"Pages in active decline: {declining_pages:,} ({pct_declining:.1f}%)")
print(f"Average age of declining pages: {avg_age_declining:.0f} days")
print(f"Average age of non-declining pages: {avg_age_stable:.0f} days")
```
